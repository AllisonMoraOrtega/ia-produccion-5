```python
import importlib
import subprocess
import sys
from pathlib import Path
import pandas as pd

required_packages = {
    'pandas': 'pandas',
    'openpyxl': 'openpyxl',
    'matplotlib': 'matplotlib',
}
```

```python
# carga planilla
codigos = Path('codigos_proyectos.xlsx')
df_codigos = pd.read_excel(codigos)

df_codigos.sample(10)
```

```python
# elimina los codigos duplicados
df_codigos = df_codigos.drop_duplicates(subset=['Codigo'], keep='first')
df_codigos.sample(10)
```

```python
# crea un nuevo excel en base a los codigos unicos
df_codigos.to_excel('codigos_unicos.xlsx', index=False)
```
