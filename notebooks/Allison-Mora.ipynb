## 2. Tablas de resumen (4)

Se generan 4 tablas de agregación sobre las dimensiones más relevantes del plan de compras.

### Tabla 1 — Monto total e ítems por Unidad de Compra

```python
tabla1_unidad_compra = (
    df.groupby('Unidad de Compra', dropna=False)
    .agg(n_items=('Nombre ítem', 'count'), monto_total=('Monto Total Ítem Año 2025', 'sum'))
    .sort_values('monto_total', ascending=False)
)
tabla1_unidad_compra
```

### Tabla 2 — Proyectos y montos por Tipo y Estado de Proyecto

```python
tabla2_tipo_estado = (
    df.groupby(['Tipo Proyecto', 'Estado Proyecto'], dropna=False)
    .agg(n_proyectos=('ID Proyecto', 'nunique'), monto_total=('Monto Total Ítem Año 2025', 'sum'))
)
tabla2_tipo_estado
```

### Tabla 3 — Top 10 responsables por monto total gestionado

```python
tabla3_top_responsables = (
    df.groupby('Nombre responsable', dropna=False)
    .agg(n_items=('Nombre ítem', 'count'), monto_total=('Monto Total Ítem Año 2025', 'sum'))
    .sort_values('monto_total', ascending=False)
    .head(10)
)
tabla3_top_responsables
```

### Tabla 4 — Resumen por Tipo de compra

```python
tabla4_tipo_compra = (
    df.groupby('Tipo compra', dropna=False)
    .agg(
        n_items=('Nombre ítem', 'count'),
        monto_promedio=('Monto Unitario Ítem', 'mean'),
        monto_total=('Monto Total Ítem Año 2025', 'sum'),
    )
    .sort_values('monto_total', ascending=False)
)
tabla4_tipo_compra
```
