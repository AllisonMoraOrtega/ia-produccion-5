## 0. Carga de librerías y datos

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', 30)
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
ARCHIVO = 'plan_de_compras_2025.xlsx'

df = pd.read_excel(ARCHIVO, sheet_name='PLAN DE COMPRAS 2025')
print('Dimensiones del dataset:', df.shape)
df.head()
```

## 6. Filtro por código de proyecto según año

El `ID Proyecto` termina con un sufijo tipo `PC25`, `PC24`, `PC23`, etc., donde los **2 últimos dígitos**
indican el año del proyecto (ej. `PC25` → año 2025). Se extrae ese sufijo y se filtra el dataset para
dejar **solo los proyectos del año 25**.

```python
df['ID Proyecto'] = df['ID Proyecto'].astype('string').str.strip()
df['Sufijo Proyecto'] = df['ID Proyecto'].str.extract(r'([A-Za-z]{2}\d{2})$')
df['Año Proyecto'] = '20' + df['Sufijo Proyecto'].str[-2:]

df['Año Proyecto'].value_counts()
```

```python
df_2025 = df[df['ID Proyecto'].str.endswith('PC25', na=False)].copy()

print(f"Registros totales: {len(df)}")
print(f"Registros año 2025 (PC25): {len(df_2025)}")

df_2025[['ID Proyecto', 'Sufijo Proyecto', 'Año Proyecto']].head(10)
```

## Resumen

- **Filtro por año:** 648 de 831 registros corresponden a proyectos del año 2025 (`PC25`).
